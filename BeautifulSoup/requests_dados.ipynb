```python
import os
import requests

from bs4 import BeautifulSoup
from dotenv import load_dotenv

load_dotenv()
```

```python
os.environ["HTTP_PROXY"] = os.getenv("HTTP_PROXY", "")
os.environ["HTTPS_PROXY"] = os.getenv("HTTPS_PROXY", "")
os.environ["NO_PROXY"] = os.getenv("NO_PROXY", "")

print("Proxy configurado.")
```

```python
url = "https://www.detran.pr.gov.br/"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
                  "AppleWebKit/537.36 (KHTML, like Gecko) "
                  "Chrome/151.0.0.0 Safari/537.36"
}

resposta = requests.get(url, headers=headers)

print(resposta.status_code)
```

```python
conteudo = resposta.text

soup = BeautifulSoup(conteudo, "html.parser")

print("HTML processado com BeautifulSoup.")
```

```python
cabecalho = soup.find("h1")

if cabecalho:
    print("Cabeçalho encontrado:")
    print(cabecalho.get_text(strip=True))
else:
    print("Nenhum h1 encontrado.")
```

```python
cabecalho = soup.find("h2")

if cabecalho:
    print("Cabeçalho encontrado:")
    print(cabecalho.get_text(strip=True))
else:
    print("Nenhum h2 encontrado.")
```

```python
link = soup.find("a", href=True)

if link:
    print("Texto:", link.get_text(strip=True))
    print("Href:", link.get("href"))
else:
    print("Nenhum link encontrado.")
```
